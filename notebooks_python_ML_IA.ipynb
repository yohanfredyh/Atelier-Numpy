# Partie 8 : Calcul scientifique

Considérons le tableau représentant les 24 températures du 1er bâtiment

1) Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la
variance et l’écart-type

```python
minimum = np.min(tb_aleatoire)
maximum = np.max(tb_aleatoire)
somme = np.sum(tb_aleatoire)
moyenne = np.mean(tb_aleatoire)
mediane = np.median(tb_aleatoire)
variance = np.var(tb_aleatoire)
ecart_type = np.std(tb_aleatoire)
print(f"minimum ={minimum}")
print(f"maximum ={maximum}")
print(f"somme ={somme}")
print(f"moyenne ={moyenne}")
print(f"mediane ={mediane}")
print(f"variance ={variance}")
print(f"ecart_type ={ecart_type}")
```

2) Standardiser le tableau des températures puis vérifier les valeurs de la moyenne et de l’écarttype
du tableau standardisé

```python
tb_standardise = (tb_aleatoire - moyenne) / ecart_type
print(tb_standardise)
print(f"moyenne standardisee ={np.mean(tb_standardise)}")
print(f"ecart_type standardise ={np.std(tb_standardise)}")
```

3) Normaliser (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé

```python
tb_normalise = (tb_aleatoire - minimum) / (maximum - minimum)
print(tb_normalise)
print(f"minimum normalise ={np.min(tb_normalise)}")
print(f"maximum normalise ={np.max(tb_normalise)}")
```

4) Convertir le tableau initial en Fahrenheit

```python
tb_fahrenheit = tb_aleatoire * 9/5 + 32
print(tb_fahrenheit)
```
