# partie 1

  1) Créer une liste Python stockant ces températures

une liste python est une structure générique. pour créer une liste sur python on utilise les [ ].

```python
temperature = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperature)
```

2) Créer un tableau NumPy équivalent stockant ces températures

pour créer un tableau numpy on utilise la fonction np.array()

```python
tableau = np.array([24.5, 25.1, 26.3, 27.0, 25.8 ])
print(f"tableau ={tableau}")
```

3) Effectuer une multiplication par 2 sur ces deux collections

pour effectuer une multiplication on utilise *

```python
multi_temps = temperature * 2
print(multi_temps)
multi_tab = tableau * 2
print(multi_tab)
```

4) Quelle différence observe-t-on ?

pour la liste les données ont étés dupliqués alors que pour le tableau chaque donnée à été multiplier par 2
