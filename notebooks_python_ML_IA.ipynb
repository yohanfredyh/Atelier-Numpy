# partie 1

  1) Créer une liste Python stockant ces températures

une liste python est une structure générique. pour créer une liste sur python on utilise les [ ].

```python
temperature = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperature)
```

2) Créer un tableau NumPy équivalent stockant ces températures

pour créer un tableau numpy on utilise la fonction np.array()

```python
tableau = np.array([24.5, 25.1, 26.3, 27.0, 25.8 ])
print(f"tableau ={tableau}")
```

3) Effectuer une multiplication par 2 sur ces deux collections

pour effectuer une multiplication on utilise *

```python
multi_temps = temperature * 2
print(multi_temps)
multi_tab = tableau * 2
print(multi_tab)
```

4) Quelle différence observe-t-on ?

pour la liste les données ont étés dupliqués alors que pour le tableau chaque donnée à été multiplier par 2

# Partie 2 – Création des données

1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro

on va utiliser np.zeros

```python
tableau_zeros = np.zeros(10)
print(tableau_zeros)
```

2) Créer un tableau contenant 10 valeurs égales à 1.

on utilise np.ones

```python
tableau_ones = np.ones(10)
print(tableau_ones)
```

3) Créer un tableau de 10 valeurs initialisées à 25.

on va utiliser np.full

```python
tableau_full = np.full((10),25)
print(tableau_full)
```

4) Créer une matrice identité de taille 5 × 5

on utilise np.eye

```python
matrice_iden = np.eye(5)
print(matrice_iden)
```

5) Créer les 24 heures (de 0 à 23) de mesure d’une journée

on va utiliser np.arange

```python
mesure_journee = np.arange(0, 24)
print(mesure_journee)
```

6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C

on utilise np.linspace

```python
valeur_repartie = np.linspace(20, 30, 10)
print(valeur_repartie)
```

7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles

on va utiliser np.random.seed

```python
np.random.seed(50)
mesures_aleatoire = np.random.uniform(10, 35, 100)
print(mesures_aleatoire)
```
